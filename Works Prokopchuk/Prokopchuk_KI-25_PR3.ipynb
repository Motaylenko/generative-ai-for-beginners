```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

token = os.environ["GITHUB_TOKEN"]
endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Покажи мені 5 рецептів для страви з наступними інгредієнтами: курка, картопля та цибуля. Для кожного рецепту перелічи всі використані інгредієнти"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "Ти корисний асистент.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Опціональні параметри
    temperature=1.,
    max_tokens=3000,
    top_p=1.    
)

print(response.choices[0].message.content)
```

```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

token = os.environ["GITHUB_TOKEN"]
endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("Кількість рецептів (наприклад, 5): ")

ingredients = input("Список інгредієнтів (наприклад, курка, картопля та морква): ")

# інтерполюємо кількість рецептів у запит та інгредієнти
prompt = f"Покажи мені {no_recipes} рецептів для страви з наступними інгредієнтами: {ingredients}. Для кожного рецепту перелічи всі використані інгредієнти"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "Ти корисний асистент.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Опціональні параметри
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
```

## Індивідуальне завдання

```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.core.credentials import AzureKeyCredential

# ✅ Додаток для аналізу та оптимізації мережевих протоколів IoT-пристроїв


def get_token() -> str:
    token = os.environ.get("GITHUB_TOKEN")
    if not token:
        raise RuntimeError("Не знайдено системну змінну середовища GITHUB_TOKEN")
    return token


def choose_protocol() -> str:
    protocol = input("Оберіть протокол (MQTT/CoAP/HTTP/WebSocket/Custom): ").strip().upper()
    valid = {"MQTT", "COAP", "HTTP", "WEBSOCKET", "CUSTOM"}
    if protocol not in valid:
        print("⚠️ Невідомий протокол, використовую MQTT за замовчуванням.")
        protocol = "MQTT"
    return protocol


def choose_traffic_profile() -> str:
    profile = input("Тип трафіку (telemetry/control/firmware_update/mixed): ").strip().lower()
    if not profile:
        profile = "telemetry"
    return profile


def build_sample_traffic(protocol: str, traffic_profile: str) -> str:
    if protocol == "MQTT":
        return f"""
[2024-12-11 10:15:00.123] CLIENT: sensor-001 CONNECT keepalive=60 clean_session=true
[2024-12-11 10:15:00.150] SERVER: CONNACK session_present=false return_code=0
[2024-12-11 10:15:00.200] CLIENT: SUBSCRIBE msg_id=1 topic="home/room1/temperature" qos=1
[2024-12-11 10:15:00.220] SERVER: SUBACK msg_id=1 qos=1
[2024-12-11 10:15:01.001] CLIENT: PUBLISH topic="home/room1/temperature" qos=1 payload="22.3" dup=false
[2024-12-11 10:15:01.010] SERVER: PUBACK msg_id=2
[2024-12-11 10:15:01.500] CLIENT: PUBLISH topic="home/room1/humidity" qos=0 payload="45.1"
[2024-12-11 10:15:30.000] CLIENT: PINGREQ
[2024-12-11 10:15:30.005] SERVER: PINGRESP
# profile={traffic_profile}
"""
    if protocol == "COAP":
        return f"""
[2024-12-11 10:15:00.050] CON: MID=0x1234, Code=0.01 (GET), Token=0xAA
 URI-Path: sensors
 URI-Path: room1
 URI-Path: temperature
[2024-12-11 10:15:00.080] ACK: MID=0x1234, Code=2.05 (Content)
 Payload: "22.3"
[2024-12-11 10:15:05.050] CON: MID=0x1235, Code=0.02 (POST), Token=0xAB
 URI-Path: actuators
 URI-Path: room1
 URI-Path: relay
 Payload: "ON"
# observe disabled, profile={traffic_profile}
"""
    if protocol == "HTTP":
        return f"""
[2024-12-11 10:15:00.010] POST /api/v1/iot/data HTTP/1.1
 Host: iot-gateway.local
 Authorization: Bearer <token>
 Content-Type: application/json
 Content-Length: 82

 {{
   "deviceId": "sensor-001",
   "type": "{traffic_profile}",
   "temperature": 22.3,
   "humidity": 45.1
 }}

[2024-12-11 10:15:00.030] HTTP/1.1 200 OK
 Content-Length: 16

 {{ "status": "ok" }}

[2024-12-11 10:15:01.100] POST /api/v1/iot/data HTTP/1.1
# інтервал відправки ≈1 c
"""
    if protocol == "WEBSOCKET":
        return f"""
[2024-12-11 10:15:00.000] WS CONNECT wss://iot-cloud.example.com/stream
[2024-12-11 10:15:00.020] WS FRAME → type=TEXT payload='{{"device":"sensor-001","event":"join","profile":"{traffic_profile}"}}'
[2024-12-11 10:15:00.040] WS FRAME ← type=TEXT payload='{{"status":"ok","session":"abc123"}}'
[2024-12-11 10:15:01.000] WS FRAME → type=TEXT payload='{{"ts":1702295701,"temp":22.3,"hum":45.1}}'
[2024-12-11 10:15:01.005] WS FRAME ← type=TEXT payload='{{"ack":true}}'
[2024-12-11 10:15:02.000] WS FRAME → type=TEXT payload='{{"ts":1702295702,"temp":22.4,"hum":45.0}}'
# RTT стабільний, втрат не видно
"""
    # CUSTOM / інше
    return f"""
[2024-12-11 10:15:00.000] FRAME src=0x01 dst=0xFF proto=CUSTOM seq=1 len=24
 payload=0x10 0x01 0x64 0x17 0x00 0x2D ...
[2024-12-11 10:15:00.050] FRAME src=0x01 dst=0xFF proto=CUSTOM seq=2 len=24
 payload=0x10 0x01 0x64 0x18 0x00 0x2C ...
[2024-12-11 10:15:00.100] FRAME src=0x01 dst=0xFF proto=CUSTOM seq=3 len=24 (RETRANSMIT)
# Є повторні передачі, можливі втрати, profile={traffic_profile}
"""


def build_prompt(protocol: str, traffic_profile: str, sample_traffic: str) -> str:
    return (
        f"Ти отримуєш фрагмент логу мережевого трафіку IoT-пристроїв.\n"
        f"Протокол: {protocol}\n"
        f"Профіль трафіку: {traffic_profile}\n\n"
        f"Фрагмент трафіку:\n{sample_traffic}\n\n"
        "Виконай детальний аналіз та оптимізацію:\n"
        "1) Опиши, як саме використовується протокол (типи повідомлень, теми/ресурси/URI, QoS/коди відповідей).\n"
        "2) Оціни частоту відправки повідомлень, розмір корисного навантаження, наявність повторних передач чи таймаутів.\n"
        "3) Визнач потенційні проблеми (затримки, перевантаження каналу, зайвий overhead, неефективні інтервали keepalive).\n"
        "4) Запропонуй конкретні кроки оптимізації (налаштування QoS, batching, агрегування даних, зміна інтервалів, стиснення тощо).\n"
        "5) Оціни базові аспекти безпеки (TLS/DTLS, автентифікація, передача чутливих даних у відкритому вигляді) "
        "та дай рекомендації.\n"
        "6) Поясни, як запропоновані зміни вплинуть на надійність, затримки та енергоспоживання IoT-пристрою."
    )


def main():
    # 1) Ініціалізація клієнта
    token = get_token()
    endpoint = "https://models.inference.ai.azure.com"
    model_name = "gpt-4o"

    client = ChatCompletionsClient(
        endpoint=endpoint,
        credential=AzureKeyCredential(token),
    )

    # 2) Ввід параметрів від користувача
    protocol = choose_protocol()
    traffic_profile = choose_traffic_profile()

    # 3) Побудова прикладу трафіку
    sample_traffic = build_sample_traffic(protocol, traffic_profile)
    print(f"\n📋 Приклад мережевого трафіку для IoT ({protocol}, profile={traffic_profile}):\n{sample_traffic}\n")

    # 4) Формування запиту до моделі
    prompt = build_prompt(protocol, traffic_profile, sample_traffic)

    # 5) Виклик моделі
    response = client.complete(
        messages=[
            {
                "role": "system",
                "content": (
                    "Ти — інженер з мережевої діагностики та оптимізації IoT-рішень. "
                    "Аналізуй логи трафіку, виявляй проблеми продуктивності та безпеки, "
                    "і пропонуй практичні рекомендації."
                ),
            },
            {
                "role": "user",
                "content": prompt,
            },
        ],
        model=model_name,
        temperature=0.7,
        max_tokens=1500,
        top_p=1.0,
    )

    print("=== АНАЛІЗ IoT-ТРАФІКУ ===\n")
    print(response.choices[0].message.content)

    # 6) Додатковий запит
    follow_up_prompt = input(
        "\nЯкщо хочеш, задай додаткове питання про оптимізацію/безпеку IoT-трафіку "
        "(або натисни Enter, щоб завершити): "
    ).strip()

    if follow_up_prompt:
        response2 = client.complete(
            messages=[
                {
                    "role": "system",
                    "content": (
                        "Ти — експерт з мережевих протоколів IoT, продуктивності та безпеки. "
                        "Давай прикладні поради та пояснення."
                    ),
                },
                {
                    "role": "user",
                    "content": follow_up_prompt,
                },
            ],
            model=model_name,
            temperature=0.7,
            max_tokens=1000,
        )
        print("\n=== ДОДАТКОВІ РЕКОМЕНДАЦІЇ ДЛЯ IoT ===\n")
        print(response2.choices[0].message.content)
    else:
        print("\n(Додатковий запит пропущено)")


if __name__ == "__main__":
    main()
```
