### Вправа 2: Перевірка налаштування ключа Github Models

Запустіть код нижче, щоб перевірити, чи правильно налаштована кінцева точка Github Models. Код просто випробовує простий базовий промпт і перевіряє завершення. Вхідний текст `о скажи, чи бачиш ти` має завершитись відповідно до контексту.


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

token = os.environ["GITHUB_TOKEN"]
endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

def get_completion(prompt, client, model_name, temperature=1.0, max_tokens=1000, top_p=1.0):
    response = client.complete(
        messages=[
            {
                "role": "system",
                "content": "Ви - корисний асистент.",
            },
            {
                "role": "user",
                "content": prompt,
            },
        ],
        model=model_name,
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p
    )
    return response.choices[0].message.content

## ---------- Виклик допоміжного методу

### 1. Встановіть основний вміст або текст промпту
text = f"""
о скажи, чи бачиш ти
"""

### 2. Використайте його в шаблоні промпту нижче
prompt = f"""
```{text}```
"""

## 3. Запустіть промпт
response = get_completion(prompt, client, model_name)
print(response)
````

### Вправа: Дослідіть свою інтуїцію
Наведені вище приклади дають вам шаблони, які ви можете використовувати для створення нових промптів (простих, складних, інструкційних тощо) - спробуйте створити інші вправи для дослідження деяких інших ідей, про які ми говорили, таких як приклади, підказки та інше.

````python
# Варіант: 9. Створіть промпт для аналізу дампів пам'яті x86/ARM процесорів.
# Мотайленко Олександр

import tiktoken

# 1️ Функція для побудови промптів
def get_memory_dump_prompts():
    base_prompt = """
    Ти — інженер з низькорівневої діагностики та реверс-інжинірингу.
    Ти отримуєш дамп пам'яті процесу або операційної системи (x86 або ARM) і маєш проаналізувати:
    • вміст стеку (stack),
    • ділянки heap,
    • області коду та даних,
    • регістри процесора, якщо вони надані.

    Проаналізуй дамп і опиши коротко:
    1. Виявлені підозрілі області або аномалії (пошкоджені кадри стеку, дивні вказівники, сигнатури шкідливого коду).
    2. Імовірну причину (переповнення буфера, use-after-free, звернення до null/invalid pointer, гонки потоків тощо).
    3. Рекомендації для подальшого налагодження (які модулі перевірити, які breakpoints поставити, які інструменти використати).
    """

    system_context = """
    Ти — аналізатор дампів пам'яті x86/ARM, який працює в режимі глибокої post-mortem діагностики.
    Твоє завдання — за вмістом пам'яті виявляти:
    • переповнення буфера (stack/heap overflow),
    • пошкодження структури heap (heap corruption),
    • використання звільненої пам'яті (use-after-free),
    • звернення до некоректних або підозрілих адрес,
    • можливі сліди шкідливого коду.

    Формат відповіді: структурований звіт з розділами:
    1) Короткий опис вхідних даних (архітектура, тип дампу, ключові адреси).
    2) Список знайдених аномалій.
    3) Аналіз можливих причин.
    4) Рекомендації для розробника / дослідника.

    Наприкінці додай короткий узагальнюючий висновок на 2–3 речення.
    """

    # прибираємо зайві пробіли по краях
    return base_prompt.strip(), system_context.strip()


# 2️ Допоміжна функція токенізації
def tokenize(text, encoding):
    tokens = encoding.encode(text)
    return tokens, len(tokens)


# 3️ Функція оцінки якості промпта
def evaluate_prompt_quality(prompt: str) -> int:
    text = prompt.lower()
    score = 0

    # Наявність аналізу
    if "аналіз" in text or "проаналізуй" in text:
        score += 2

    # Наявність помилок / аномалій / підозрілих областей
    if ("помилки" in text) or ("аномалії" in text) or ("підозрілі" in text):
        score += 2

    # Наявність рекомендацій
    if "рекомендації" in text:
        score += 2

    # Достатня довжина
    if len(prompt.split()) > 30:
        score += 1

    # Вимога структурованого звіту
    if "структурований" in text or "звіт" in text:
        score += 1

    return score


# 4️ Основна логіка скрипта
def main():
    # Створюємо промпти
    base_prompt, system_context = get_memory_dump_prompts()

    # Обираємо кодування токенів
    encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")

    # Токенізація
    base_tokens, base_len = tokenize(base_prompt, encoding)
    context_tokens, context_len = tokenize(system_context, encoding)

    print("Токени базового промпта:", base_tokens)
    print("Токени системного контексту:", context_tokens)
    print(f"Кількість токенів у базовому промпті: {base_len}")
    print(f"Кількість токенів у системному контексті: {context_len}")

    # Оцінка якості промптів
    base_score = evaluate_prompt_quality(base_prompt)
    context_score = evaluate_prompt_quality(system_context)

    print("\nОцінка якості промптів:")
    print(f"Базовий промпт: {base_score}/8 балів")
    print(f"Системний контекст: {context_score}/8 балів")

    # Пропозиції щодо покращення
    improvements = [
        "Додати приклади реальних дампів пам'яті (фрагменти стеку, heap, таблиць сторінок).",
        "Явно вказати формат виводу (Markdown або JSON з полями type / anomalies / root_cause / recommendations).",
        "Додати параметр архітектури та формату дампу (x86, x86_64, ARMv7, ARM64, core dump, minidump).",
        "Ввести рівні деталізації аналізу: короткий (summary), детальний (detailed), forensics-mode.",
        "Додати контекст про процес/бінарник (ім'я процесу, версія збірки, наявні символи) для точніших висновків.",
    ]

    print("\nПропозиції для покращення промптів:")
    for item in improvements:
        print("-", item)


if __name__ == "__main__":
    main()
````
