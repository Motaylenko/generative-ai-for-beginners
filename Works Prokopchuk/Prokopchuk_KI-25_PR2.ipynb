### Вправа 2: Перевірка налаштування ключа Github Models

Запустіть код нижче, щоб перевірити, чи правильно налаштована кінцева точка Github Models. Код просто випробовує простий базовий промпт і перевіряє завершення. Вхідний текст `о скажи, чи бачиш ти` має завершитись відповідно до контексту.


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

token = os.environ["GITHUB_TOKEN"]
endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

def get_completion(prompt, client, model_name, temperature=1.0, max_tokens=1000, top_p=1.0):
    response = client.complete(
        messages=[
            {
                "role": "system",
                "content": "Ви - корисний асистент.",
            },
            {
                "role": "user",
                "content": prompt,
            },
        ],
        model=model_name,
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p
    )
    return response.choices[0].message.content

## ---------- Виклик допоміжного методу

### 1. Встановіть основний вміст або текст промпту
text = f"""
о скажи, чи бачиш ти
"""

### 2. Використайте його в шаблоні промпту нижче
prompt = f"""
```{text}```
"""

## 3. Запустіть промпт
response = get_completion(prompt, client, model_name)
print(response)
````

### Вправа: Дослідіть свою інтуїцію
Наведені вище приклади дають вам шаблони, які ви можете використовувати для створення нових промптів (простих, складних, інструкційних тощо) - спробуйте створити інші вправи для дослідження деяких інших ідей, про які ми говорили, таких як приклади, підказки та інше.

````python
# Варіант: 9. Створіть промпт для аналізу та оптимізації мережевих протоколів IoT-пристроїв.
# Мотайленко Олександр

import tiktoken


# 1️ Побудова промптів
def build_iot_protocol_prompts():
    user_prompt = """
    Ти виступаєш у ролі експерта з мережевих технологій для IoT.
    На вхід ти отримуєш дампи мережевого трафіку (файли pcap, журнали роботи шлюзів,
    брокера повідомлень або хмарного сервісу) в інфраструктурі з великою кількістю IoT-пристроїв.

    Твоє завдання — проаналізувати трафік і з’ясувати:
    • які мережеві протоколи та стек використовуються (MQTT, CoAP, HTTP(S), WebSocket, власні бінарні протоколи тощо);
    • як виглядають шаблони обміну даними між сенсорами, актуаторами, шлюзами та хмарою;
    • які налаштування застосовано (QoS, інтервали публікацій, розмір корисного навантаження, частота повторних відправлень);
    • чи є проблеми з затримками, втратою пакетів або перевантаженням каналу;
    • чи не передаються конфіденційні дані у відкритому вигляді та чи використовується шифрування/автентифікація.

    Сформуй короткий, але інформативний опис:
    1. Виявлених проблем і нетипових ситуацій (надмірний обсяг телеметрії, занадто часті keep-alive,
       великі або «шумні» повідомлення, повторні відправлення, відсутність захисту каналу).
    2. Можливих причин цих проблем (невдалий вибір протоколу для даного середовища,
       некоректно задані таймери, відсутність агрегації даних, неефективний формат повідомлень чи топологія).
    3. Практичних кроків з оптимізації:
       • на рівні протоколів (налаштування QoS, підтверджень, retry-механізмів, вибір протоколу),
       • на рівні трафіку (агрегація, стиснення, кешування, зміна частоти надсилання),
       • на рівні безпеки (шифрування, керування ключами, токени доступу, сертифікати).
    """

    system_prompt = """
    Ти — спеціалізований аналізатор мережевих протоколів для IoT-систем,
    який працює у режимі розширеної діагностики продуктивності та безпеки.

    На основі дампів трафіку, логів брокера та показників мережі ти маєш:
    • знаходити перевантаження каналів і потенційні «вузькі місця»;
    • виявляти зайві, дубльовані або занадто часті повідомлення;
    • визначати помилки та зловживання параметрами QoS, keep-alive, timeout, retry;
    • помічати використання незахищених протоколів або передачу чутливих даних без шифрування;
    • ідентифікувати пристрої, що поводяться аномально (часті reconnection, flood-подібний трафік, flapping).

    Відповідь потрібно оформити як структурований звіт із такими розділами:
    1) Опис вхідних даних:
       • формат джерела (pcap, логи брокера, журнали шлюзу),
       • виявлені протоколи, порти та основні учасники обміну,
       • короткий опис топології (IoT-пристрій → шлюз → брокер/хмара).
    2) Перелік знайдених аномалій та потенційних проблем
       з конкретними прикладами (IP-адреси, теми MQTT, URI, порти).
    3) Пояснення ймовірних причин (конфігурація пристроїв, особливості каналу, обмеження апаратури).
    4) Рекомендації для розробника або архітектора:
       • які параметри варто змінити (частота відправлень, QoS, timeout, розмір повідомлень),
       • чи доцільно замінити або комбінувати протоколи,
       • як посилити захист трафіку,
       • які інструменти використовувати для додаткового аналізу (фільтри Wireshark, спеціалізовані IoT-аналізатори, тестові стенди).

    Заверши звіт коротким висновком (2–3 речення), у якому підкресли,
    які зміни дадуть найбільший приріст у надійності та ефективності роботи IoT-мережі.
    """

    return user_prompt.strip(), system_prompt.strip()


# 2️ Обгортка над tiktoken для підрахунку токенів
def encode_text_to_tokens(text: str, encoding) -> tuple[list[int], int]:
    encoded = encoding.encode(text)
    return encoded, len(encoded)


# 3️ Проста «оцінка якості» промпта
def score_prompt(prompt: str) -> int:
    lowered = prompt.lower()
    points = 0

    # Чи є згадки про аналіз
    if "аналіз" in lowered or "проаналізуй" in lowered:
        points += 2

    # Чи згадуються помилки / аномалії / проблеми
    if ("помилки" in lowered) or ("аномалії" in lowered) or ("проблем" in lowered):
        points += 2

    # Чи проситься надати рекомендації
    if "рекомендації" in lowered:
        points += 2

    # Орієнтовна достатня довжина
    if len(prompt.split()) > 30:
        points += 1

    # Вимога видати звіт
    if "структурований" in lowered or "звіт" in lowered:
        points += 1

    return points


# 4️ Основна функція запуску
def main():
    user_prompt, system_prompt = build_iot_protocol_prompts()

    # вибираємо кодування для обчислення кількості токенів
    encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")

    # рахуємо токени
    user_tokens, user_len = encode_text_to_tokens(user_prompt, encoding)
    system_tokens, system_len = encode_text_to_tokens(system_prompt, encoding)

    print("=== Інформація про токени ===")
    print("К-сть токенів у user-промпті:", user_len)
    print("К-сть токенів у system-промпті:", system_len)

    # оцінюємо якість текстів
    user_score = score_prompt(user_prompt)
    system_score = score_prompt(system_prompt)

    print("\n=== Оцінка якості промптів (0–8) ===")
    print(f"user-промпт:   {user_score}/8")
    print(f"system-промпт: {system_score}/8")

    # ідеї для подальшого покращення
    suggestions = [
        "Додати приклади реальних записів трафіку (фрагменти pcap, уривки логів брокера).",
        "Передбачити опцію виводу у форматі JSON з полями protocol_summary, issues, root_causes, recommendations.",
        "Уточнити обмеження середовища (тип каналів: Wi-Fi, LTE, LoRaWAN, наявність батарейного живлення).",
        "Додати прапорець режиму аналізу: performance-focus, security-focus, mixed.",
        "Надати можливість передавати список типових пристроїв (датчики, актуатори, камери), щоб робити точніші поради.",
    ]

    print("\n=== Ідеї для вдосконалення промпта ===")
    for s in suggestions:
        print("-", s)


if __name__ == "__main__":
    main()
````
